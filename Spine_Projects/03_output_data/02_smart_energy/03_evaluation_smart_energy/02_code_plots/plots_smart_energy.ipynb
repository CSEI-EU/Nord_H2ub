```python
#waterfall
# -----------------------
# Final Categories
# -----------------------
categories = [
    "Electrolyser",       # investment
    "Methanol",           # investment
    "Other Units",        # investment
    "Storage",            # investment
    "Power",              # variable
    "Grid tariffs",       # variable
    "CO$_2$ + Water",        # variable
    "Other",              # variable
    "DH",                 # revenue
    "O$_2$",                 # revenue
    "LCOM"                # total
]

# -----------------------
# Values (adjust O2 if needed)
# -----------------------
steps = np.array([
    92.44,    # Electrolyser
    66.18,    # Methanol
    38.9,     # Other Units
    1,      # Storage
    818.96,    # Power
    137.81,     # Grid tariffs
    41.12,     # CO2 + Water
    52,    # Other
    -17.20,   # DH revenue
    -262.90,   # O2 revenue
    968.65    # LCOM
], dtype=float)

intermediate = steps[:-1]
total_LCOM = steps[-1]

# cumulative stacking
cum_start = np.concatenate([[0], np.cumsum(intermediate)])[:-1]
cum_end = cum_start + intermediate

# -----------------------
# Colors
# -----------------------
color_invest = "#2f4fb3"     # dark blue
color_variable = "#7ea9ff"   # light blue
color_neg = "#2aa68a"        # green
color_total = "#bdbdbd"      # grey

# Investment indices (0–3)
investment_indices = [0,1,2,3]
# Variable indices (4–7)
variable_indices = [4,5,6,7]

# -----------------------
# Plot
# -----------------------
fig, ax = plt.subplots(figsize=(16,8))
bar_width = 0.75
n_inter = len(intermediate)

for i, (start, end, val) in enumerate(zip(cum_start, cum_end, intermediate)):

    if val >= 0:
        if i in investment_indices:
            color = color_invest
        else:
            color = color_variable
        bottom = start
        height = val
    else:
        color = color_neg
        bottom = end
        height = -val

    ax.bar(i, height, bottom=bottom, width=bar_width,
           color=color, zorder=3)

    # annotation
    y_annot = bottom + height
    txt = f"{int(val):+d}" if val < 0 else f"{int(val)}"
    ax.text(i, y_annot + (max(intermediate) * 0.02),
            txt,
            ha='center', va='bottom',
            fontsize=12, fontweight='bold',
            color=color)

# connectors
for i in range(n_inter - 1):
    x0 = i + bar_width/2
    x1 = i + 1 - bar_width/2
    y = cum_end[i]
    ax.plot([x0, x1], [y, y], color='black', lw=1.0)

# final total bar
total_index = n_inter
ax.bar(total_index, total_LCOM, width=bar_width,
       color=color_total, zorder=3)
ax.text(total_index,
        total_LCOM + (max(intermediate) * 0.02),
        f"{int(total_LCOM)}",
        ha='center', va='bottom',
        fontsize=12, fontweight='bold',
        color='#666666')

# -----------------------
# Axis formatting
# -----------------------
ax.set_xticks(np.arange(len(categories)))
ax.set_xticklabels(categories,
                   rotation=55,
                   ha='right',
                   fontsize=14,
                   fontweight='bold')

ax.set_ylabel('[€/t Methanol]', fontsize=18, fontweight='bold')

for label in ax.get_yticklabels():
    label.set_fontsize(14)
    label.set_fontweight('bold')

ax.yaxis.grid(True, linestyle='-', linewidth=0.5, color='#e9e9e9')

# -----------------------
# Legend (outside, square patches)
# -----------------------
legend_handles = [
    Patch(facecolor=color_invest, label='Investment Costs'),
    Patch(facecolor=color_variable, label='Variable Costs'),
    Patch(facecolor=color_neg, label='Revenue'),
    Patch(facecolor=color_total, label='LCOM')
]

leg = ax.legend(handles=legend_handles,
                loc='center left',
                bbox_to_anchor=(1.02, 0.5),
                fontsize=14,
                frameon=False)

for text in leg.get_texts():
    text.set_fontweight('bold')

plt.tight_layout(rect=[0, 0, 0.88, 1])
plt.savefig("fig-waterfall.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
#tornado plot
# -----------------------
# Base value
# -----------------------
base = 968.65

# -----------------------
# Sensitivity data
# (value at -5%, value at +5%)
# -----------------------
sens_5 = {
    "Demand": (961.60, 974.77),
    "WACC": (961.92, 975.46),
    "Lifetime": (972.13, 965.59),
    "Investment Costs": (985.36, 979.06),
    "DH Price": (969.51, 967.79),
    "CO$_2$ Price": (966.75, 970.54),
    "Power Price": (923.08, 1045.36),
    "Efficiency": (1023.52, 915.42),
    "Grid Tariff": (961.53, 975.73),
    "Water Price": (968.48, 968.81),
    "O$_2$ Price": (981.79, 955.50),
    "O$_2$ Volume": (981.79, 955.50),
    "PPA Price": (967.71, 969.54),
}

# -----------------------
# Sensitivity data
# (value at -10%, value at +10%)
# -----------------------
sens_10 = {
    "Demand": (962.16, 1034.68),
    "WACC": (955.29, 982.36),
    "Lifetime": (976.11, 962.89),
    "Investment Costs": (1006.85, 988.65),
    "DH Price": (970.37, 966.93),
    "CO$_2$ Price": (964.86, 972.43),
    "Power Price": (880.98, 1119.04),
    "Efficiency": (1113.33, 873.90),
    "Grid Tariff": (953.16, 982.80),
    "Water Price": (968.32, 968.97),
    "O$_2$ Price": (994.94, 942.36),
    "O$_2$ Volume": (994.94, 942.36),
    "PPA Price": (966.77, 970.46),
}

# -----------------------
# Cost type groups
# -----------------------
fixed_costs = {
    "WACC",
    "Lifetime",
    "Investment Costs",
    "Demand",
    "Efficiency",
}

variable_costs = {
    "DH Price",
    "CO$_2$ Price",
    "Power Price",
    "Grid Tariff",
    "Water Price",
    "O$_2$ Price",
    "O$_2$ Volume",
    "PPA Price",
}

# -----------------------
# Colors
# -----------------------
fixed_dark = "#2f4fb3"
fixed_light = "#a8b8f0"

var_dark = "#2aa68a"
var_light = "#8fd6b5"
color_base = "black"

# -----------------------
# Common x-limits
# -----------------------
all_values = [base]
for d in (sens_5, sens_10):
    for minus, plus in d.values():
        all_values.extend([minus, plus])

x_min = min(all_values)
x_max = max(all_values)
pad = (x_max - x_min) * 0.06
common_xlim = (x_min - pad, x_max + pad)

# -----------------------
# Ordering: largest total span first
# -----------------------
items = []
for param in sens_10.keys():
    m5, p5 = sens_5[param]
    m10, p10 = sens_10[param]

    left5, right5 = min(m5, p5), max(m5, p5)
    left10, right10 = min(m10, p10), max(m10, p10)

    span = max(right5 - left5, right10 - left10)
    fam = "fixed" if param in fixed_costs else "variable"

    items.append((param, span, fam))

items.sort(key=lambda x: x[1], reverse=True)
ordered_params = [x[0] for x in items]

# -----------------------
# Plot
# -----------------------
fig, ax = plt.subplots(figsize=(14, 9))

bar_h = 0.62
y = np.arange(len(ordered_params))

def draw_one_bar(ax, y_pos, left5, right5, left10, right10, family):
    if family == "fixed":
        dark = fixed_dark
        light = fixed_light
    else:
        dark = var_dark
        light = var_light

    # 10% range (background)
    ax.barh(
        y_pos,
        right10 - left10,
        left=left10,
        height=0.40,
        color=light,
        alpha=0.8,
        edgecolor="none",
        zorder=1,
    )

    # 5% range (foreground)
    ax.barh(
        y_pos,
        right5 - left5,
        left=left5,
        height=0.30,
        color=dark,
        edgecolor="none",
        zorder=2,
    )

# Draw all bars
for i, param in enumerate(ordered_params):
    m5, p5 = sens_5[param]
    m10, p10 = sens_10[param]

    left5, right5 = min(m5, p5), max(m5, p5)
    left10, right10 = min(m10, p10), max(m10, p10)

    family = "fixed" if param in fixed_costs else "variable"
    draw_one_bar(ax, i, left5, right5, left10, right10, family)

# Base line
ax.axvline(base, color=color_base, linestyle="--", linewidth=1.5, zorder=4)

# Axes
ax.set_yticks(y)
ax.set_yticklabels(ordered_params)
ax.invert_yaxis()
ax.set_xlabel("LCOA [€/t Methanol]")
ax.set_xlim(common_xlim)
ax.grid(axis="x", linestyle=":", alpha=0.5)

# Legend
legend_handles = [
    Patch(facecolor=fixed_dark, edgecolor="none", label="Fixed Costs (±5%)"),
    Patch(facecolor=fixed_light, edgecolor="none", label="Fixed Costs (±10%)"),
    Patch(facecolor=var_dark, edgecolor="none", label="Variable Costs (±5%)"),
    Patch(facecolor=var_light, edgecolor="none", label="Variable Costs (±10%)"),
]

fig.legend(
    handles=legend_handles,
    loc="lower center",
    ncol=2,
    frameon=False,
    bbox_to_anchor=(0.5, -0.02),
)

plt.tight_layout(rect=[0, 0.07, 1, 1])

plt.savefig(
    "fig-tornado_plot.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white",
)

plt.show()
```

```python
#relationship plot
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

# -----------------------
# Data
# -----------------------
data = [
    # Year, Scenario, LCOM [€/t Methanol], Methanol Production [t/a], Electrolyzer Capacity [MW]

    ("SQ",   "Base Case",              1037.83,   52350.15,   86.25),
    ("SQ",   "Wind Inv",               1013.68,   52350.15,   83.43),
    ("SQ",   "RES",                   1432.28,   52350.15,  112.36),
    ("SQ",   "Local RES + Wind Inv",   1427.33,   52350.15,  111.92),
    ("SQ",   "Desalination",           1059.07,   52350.15,   83.90),

    ("2035", "Base Case",               924.43,  132445.63,  267.79),
    ("2035", "Wind Inv",                905.57,  132445.63,  267.50),
    ("2035", "RES",                     891.71,  132445.63,  315.04),
    ("2035", "Local RES + Wind Inv",    892.59,  132445.63,  315.03),
    ("2035", "Desalination",            947.22,  132445.63,  267.79),

    ("2045", "Base Case",               968.65,  326145.53,  525.44),
    ("2045", "Wind Inv",                942.96,  326145.53,  439.88),
    ("2045", "RES",                     817.49,  326145.53,  702.11),
    ("2045", "Local RES + Wind Inv",    817.49,  326145.53,  702.11),
    ("2045", "Desalination",           1019.30,  326145.53,  561.05),
]

# -----------------------
# Colors by scenario
# -----------------------
colors = {
    "Base Case": "#2f4fb3",
    "Wind Inv": "#7ea9ff",
    "RES": "#2aa68a",
    "Local RES + Wind Inv": "#8fd6b5",
    "Desalination": "#f0715f",
}

# -----------------------
# Markers by year
# -----------------------
markers = {
    "SQ": "o",
    "2035": "s",
    "2045": "^",
}

# -----------------------
# Bubble size scaling
# Stronger contrast between production values
# -----------------------
def bubble_size(methanol_production):
    return 18 * (methanol_production / 50000.0) ** 2.1

# -----------------------
# Jitter overlapping points
# -----------------------
def add_jitter(points, x_tol=3.0, y_tol=6.0):
    offsets = {}
    used = set()
    groups = []

    for i, p in enumerate(points):
        if i in used:
            continue

        _, _, y, _, x = p
        group = [i]
        used.add(i)

        for j, q in enumerate(points[i + 1:], start=i + 1):
            if j in used:
                continue

            _, _, y2, _, x2 = q
            if abs(x2 - x) <= x_tol and abs(y2 - y) <= y_tol:
                group.append(j)
                used.add(j)

        groups.append(group)

    for group in groups:
        n = len(group)

        if n == 1:
            offsets[group[0]] = (0.0, 0.0)
            continue

        xs = np.linspace(-4.0, 4.0, n)
        ys = np.linspace(-8.0, 8.0, n)

        for k, idx in enumerate(group):
            offsets[idx] = (xs[k], ys[k])

    return offsets

# -----------------------
# Figure
# -----------------------
fig, ax = plt.subplots(figsize=(11, 8.8))

offsets = add_jitter(data)

# Scatter bubbles
for idx, (year, scenario, lcom, production, capacity) in enumerate(data):
    dx, dy = offsets[idx]

    ax.scatter(
        capacity + dx,
        lcom + dy,
        s=bubble_size(production),
        color=colors[scenario],
        marker=markers[year],
        edgecolors="black",
        linewidths=0.6,
        alpha=0.88,
        zorder=3
    )

# -----------------------
# Axis formatting
# -----------------------
ax.set_xlabel("Electrolyzer Capacity [MW]", fontsize=15, fontweight="bold")
ax.set_ylabel("LCOM [€/t Methanol]", fontsize=15, fontweight="bold")

for label in ax.get_xticklabels():
    label.set_fontsize(12)
    label.set_fontweight("bold")

for label in ax.get_yticklabels():
    label.set_fontsize(12)
    label.set_fontweight("bold")

ax.grid(True, linestyle="-", linewidth=0.5, color="#e9e9e9", zorder=0)

# -----------------------
# Legend handles
# -----------------------
scenario_handles = [
    Line2D(
        [0], [0],
        marker='o',
        color='w',
        markerfacecolor=color,
        markeredgecolor='black',
        markeredgewidth=0.5,
        markersize=10,
        label=name
    )
    for name, color in colors.items()
]

year_handles = [
    Line2D(
        [0], [0],
        marker=marker,
        color='black',
        linestyle='',
        markersize=10,
        label=year
    )
    for year, marker in markers.items()
]

size_handles = [
    Line2D(
        [0], [0],
        marker='o',
        color='black',
        markerfacecolor='white',
        markeredgewidth=0.8,
        linestyle='',
        markersize=s
    )
    for s in [8, 16, 26]
]

# -----------------------
# Column titles (aligned)
# -----------------------
fig.text(0.18, 0.11, "Scenario", ha="center", fontsize=11, fontweight="bold")
fig.text(0.50, 0.11, "Year", ha="center", fontsize=11, fontweight="bold")
fig.text(0.82, 0.11, "Methanol production [t/a]", ha="center", fontsize=11, fontweight="bold")

# -----------------------
# Legends (without titles)
# -----------------------
fig.legend(
    handles=scenario_handles,
    loc="lower center",
    bbox_to_anchor=(0.18, -0.03),
    frameon=False,
    fontsize=10
)

fig.legend(
    handles=year_handles,
    loc="lower center",
    bbox_to_anchor=(0.50, -0.03),
    frameon=False,
    fontsize=10
)

fig.legend(
    size_handles,
    ["50,000", "150,000", "300,000"],
    loc="lower center",
    bbox_to_anchor=(0.82, -0.03),
    frameon=False,
    labelspacing=1.4,
    handletextpad=1.2
)

# -----------------------
# Spines
# -----------------------
for spine in ax.spines.values():
    spine.set_linewidth(1.0)

plt.tight_layout(rect=[0, 0.12, 1, 1])

# -----------------------
# Save
# -----------------------
plt.savefig(
    "fig-lcom_capacity_production.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()
```
